```python
import random
import math

def get_distance_matrix(nodes):
    dist = {}
    print("Enter distance between nodes (enter 'inf' for no direct edge):\n")
    for i in range(len(nodes)):
        for j in range(i+1, len(nodes)):
            x, y = nodes[i], nodes[j]
            val = input(f"Distance {x} -> {y}: ")
            if val.strip().lower() == "inf":
                w = math.inf
            else:
                w = float(val)
            dist[(x,y)] = w
            dist[(y,x)] = w
    return dist

random.seed(42)
nodes = ['a','b','c','d','e','f','g','h']
dist = get_distance_matrix(nodes)

n = len(nodes)
index = {nodes[i]:i for i in range(n)}
pheromone = [[1.0 if dist.get((nodes[i],nodes[j]),math.inf)!=math.inf and i!=j else 0.0
              for j in range(n)] for i in range(n)]

def distance(path):
    s = 0
    for i in range(len(path)-1):
        d = dist.get((path[i],path[i+1]),math.inf)
        if d==math.inf: return math.inf
        s += d
    return s

def heuristic(i,j):
    d = dist.get((nodes[i],nodes[j]),math.inf)
    return 1.0/d if d!=math.inf and d!=0 else 0.0

def choose_next(current, visited, alpha, beta):
    i = current
    probs = []
    denom = 0.0
    for j in range(n):
        if j in visited:
            probs.append(0.0); continue
        if dist.get((nodes[i],nodes[j]),math.inf)==math.inf:
            probs.append(0.0); continue

        val = (pheromone[i][j] ** alpha) * (heuristic(i,j) ** beta)
        probs.append(val)
        denom += val

    if denom == 0:
        choices = [j for j in range(n) if j not in visited
                   and dist.get((nodes[i],nodes[j]),math.inf)!=math.inf]
        return random.choice(choices) if choices else None

    r = random.random()
    cum = 0.0
    for j,p in enumerate(probs):
        cum += p/denom
        if r <= cum:
            return j

    for j in range(n):
        if probs[j]>0: return j
    return None

def run_aco(start, goal, ants=40, iterations=120, alpha=1.0, beta=6.0, rho=0.3, q=120.0):
    best_path = None
    best_cost = math.inf

    for _ in range(iterations):
        all_paths = []
        for _ in range(ants):
            path = [start]
            visited = {start}
            current = start

            while current != goal:
                nxt = choose_next(current, visited, alpha, beta)
                if nxt is None: break
                path.append(nxt)
                visited.add(nxt)
                current = nxt

            if path[-1] != goal:
                continue

            p = [nodes[x] for x in path]
            c = distance(p)

            if c < best_cost:
                best_cost = c
                best_path = p

            all_paths.append((path,c))

        for i in range(n):
            for j in range(n):
                pheromone[i][j] *= (1 - rho)
                if pheromone[i][j] < 1e-8:
                    pheromone[i][j] = 1e-8

        for path,cost in all_paths:
            if cost == math.inf: continue
            dep = q / cost
            for k in range(len(path)-1):
                i,j = path[k], path[k+1]
                pheromone[i][j] += dep
                pheromone[j][i] += dep

    return best_path, best_cost


start_node = index['a']
goal_node = index['h']

best_path, best_cost = run_aco(start_node, goal_node)

if best_path is None:
    print("No feasible route found between a and h.")
else:
    print("BEST ROUTE FOUND\n")
    print("Route :", " -> ".join(best_path))
    print("Cost  :", best_cost)

```
